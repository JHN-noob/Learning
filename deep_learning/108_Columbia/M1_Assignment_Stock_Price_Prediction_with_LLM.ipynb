```python
# Step 2 to Step 5: Feature Engineering, Modeling, Evaluation, and Strategy Discussion

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# --- Step 2: Feature Engineering ---
df = data.copy()

# Flatten columns if yfinance returns a MultiIndex
if isinstance(df.columns, pd.MultiIndex):
    df.columns = [col[0] for col in df.columns]

# Keep only needed columns and drop missing values
df = df[['Open', 'High', 'Low', 'Close', 'Volume']].dropna().copy()

# Basic return features
df['Return_1d'] = df['Close'].pct_change()
df['Return_5d'] = df['Close'].pct_change(5)

# Moving averages
df['MA_5'] = df['Close'].rolling(5).mean()
df['MA_10'] = df['Close'].rolling(10).mean()
df['MA_20'] = df['Close'].rolling(20).mean()

# Rolling volatility
df['Volatility_5'] = df['Return_1d'].rolling(5).std()
df['Volatility_10'] = df['Return_1d'].rolling(10).std()

# Momentum
df['Momentum_5'] = df['Close'] - df['Close'].shift(5)
df['Momentum_10'] = df['Close'] - df['Close'].shift(10)

# Price range / intraday structure
df['High_Low_Spread'] = df['High'] - df['Low']
df['Open_Close_Spread'] = df['Open'] - df['Close']

# Volume features
df['Volume_Change'] = df['Volume'].pct_change()
df['Volume_MA_5'] = df['Volume'].rolling(5).mean()

# RSI(14)
delta = df['Close'].diff()
gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)
avg_gain = gain.rolling(14).mean()
avg_loss = loss.rolling(14).mean()
rs = avg_gain / avg_loss
df['RSI_14'] = 100 - (100 / (1 + rs))

# MACD
ema_12 = df['Close'].ewm(span=12, adjust=False).mean()
ema_26 = df['Close'].ewm(span=26, adjust=False).mean()
df['MACD'] = ema_12 - ema_26
df['MACD_Signal'] = df['MACD'].ewm(span=9, adjust=False).mean()

# Bollinger Bands width
rolling_mean_20 = df['Close'].rolling(20).mean()
rolling_std_20 = df['Close'].rolling(20).std()
df['Bollinger_Upper'] = rolling_mean_20 + 2 * rolling_std_20
df['Bollinger_Lower'] = rolling_mean_20 - 2 * rolling_std_20
df['Bollinger_Width'] = df['Bollinger_Upper'] - df['Bollinger_Lower']

# Time features
df['DayOfWeek'] = df.index.dayofweek
df['Month'] = df.index.month

# Target: next day's closing price
df['Target_Next_Close'] = df['Close'].shift(-1)

# Clean final dataset
df_model = df.dropna().copy()

print("Engineered dataset shape:", df_model.shape)
print("\nFeature columns:")
feature_cols = [c for c in df_model.columns if c != 'Target_Next_Close']
print(feature_cols)

# --- Step 3: Train/Test Split ---
# Use time-based split, not random split, because this is time series data.
split_idx = int(len(df_model) * 0.8)

train_df = df_model.iloc[:split_idx].copy()
test_df = df_model.iloc[split_idx:].copy()

X_train = train_df[feature_cols]
y_train = train_df['Target_Next_Close']
X_test = test_df[feature_cols]
y_test = test_df['Target_Next_Close']

print("\nTrain shape:", X_train.shape, y_train.shape)
print("Test shape:", X_test.shape, y_test.shape)

# --- Step 4: Build Models ---
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)
lin_pred = lin_reg.predict(X_test)

rf_reg = RandomForestRegressor(
    n_estimators=300,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)
rf_reg.fit(X_train, y_train)
rf_pred = rf_reg.predict(X_test)

# --- Evaluation helper ---
def evaluate_model(name, y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    print(f"\n{name} Performance")
    print("-" * 40)
    print(f"RMSE: {rmse:.4f}")
    print(f"MAE : {mae:.4f}")
    print(f"R²  : {r2:.4f}")
    return {"Model": name, "RMSE": rmse, "MAE": mae, "R2": r2}

results = []
results.append(evaluate_model("Linear Regression", y_test, lin_pred))
results.append(evaluate_model("Random Forest", y_test, rf_pred))

results_df = pd.DataFrame(results).sort_values(by="RMSE")
print("\nSummary of model performance:")
print(results_df)

best_model_name = results_df.iloc[0]["Model"]
best_pred = rf_pred if best_model_name == "Random Forest" else lin_pred

# --- Feature importance for Random Forest ---
rf_importance = pd.DataFrame({
    "Feature": feature_cols,
    "Importance": rf_reg.feature_importances_
}).sort_values(by="Importance", ascending=False)

print("\nTop 10 Random Forest feature importances:")
print(rf_importance.head(10))

# --- Visualizations ---
plt.figure(figsize=(12, 5))
plt.plot(y_test.index, y_test.values, label='Actual Next Close')
plt.plot(y_test.index, lin_pred, label='Linear Regression Prediction', alpha=0.8)
plt.plot(y_test.index, rf_pred, label='Random Forest Prediction', alpha=0.8)
plt.title('Actual vs Predicted Next-Day Closing Price')
plt.xlabel('Date')
plt.ylabel('Price')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
plt.barh(rf_importance['Feature'].head(10)[::-1], rf_importance['Importance'].head(10)[::-1])
plt.title('Top 10 Feature Importances (Random Forest)')
plt.xlabel('Importance')
plt.tight_layout()
plt.show()

# --- Step 5: Simple Trading Strategy Discussion ---
# Strategy idea:
# If predicted next-day close > today's close => go long for 1 day
# Else => stay in cash
strategy_df = test_df[['Close', 'Target_Next_Close']].copy()
strategy_df['Predicted_Next_Close'] = best_pred
strategy_df['Signal'] = (strategy_df['Predicted_Next_Close'] > strategy_df['Close']).astype(int)

# Actual next-day return relative to today's close
strategy_df['Actual_Next_Day_Return'] = (
    strategy_df['Target_Next_Close'] - strategy_df['Close']
) / strategy_df['Close']

# Strategy return: earn next-day return only when signal == 1
strategy_df['Strategy_Return'] = strategy_df['Signal'] * strategy_df['Actual_Next_Day_Return']
strategy_df['Buy_and_Hold_Return'] = strategy_df['Actual_Next_Day_Return']

strategy_df['Cumulative_Strategy'] = (1 + strategy_df['Strategy_Return']).cumprod()
strategy_df['Cumulative_Buy_Hold'] = (1 + strategy_df['Buy_and_Hold_Return']).cumprod()

print("\nStrategy summary")
print("-" * 40)
print("Best prediction model used for signal generation:", best_model_name)
print("Number of long signals:", int(strategy_df['Signal'].sum()))
print("Average strategy daily return:", strategy_df['Strategy_Return'].mean())
print("Average buy-and-hold daily return:", strategy_df['Buy_and_Hold_Return'].mean())
print("Final cumulative strategy return:", strategy_df['Cumulative_Strategy'].iloc[-1] - 1)
print("Final cumulative buy-and-hold return:", strategy_df['Cumulative_Buy_Hold'].iloc[-1] - 1)

plt.figure(figsize=(12, 5))
plt.plot(strategy_df.index, strategy_df['Cumulative_Strategy'], label='Strategy')
plt.plot(strategy_df.index, strategy_df['Cumulative_Buy_Hold'], label='Buy & Hold')
plt.title('Trading Strategy vs Buy-and-Hold')
plt.xlabel('Date')
plt.ylabel('Cumulative Return (Growth of $1)')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Discussion points
print("\nDiscussion")
print("-" * 40)
print("""
1. This is a simple educational strategy, not an investment recommendation.
2. Stock prediction is noisy, and next-day prices are difficult to forecast reliably.
3. Random Forest often captures nonlinear effects better than linear regression.
4. Even if prediction metrics look decent, trading performance can still be weak after
   transaction costs, slippage, taxes, and market regime changes.
5. For real trading, use walk-forward validation, out-of-sample testing, transaction-cost
   modeling, and risk management rules.
""")
```

- **Action**: Buy stocks when the model predicts an increase (class 1).

#### Contrarian Strategy:

- **Focus**: Take advantage of the model's high recall for no price increase predictions (class 0).
- **Action**: Sell short or avoid buying stocks on days when the model predicts no increase.
- **Risk Management**: This strategy assumes that the model’s predictions of no price increase are more reliable. It requires a careful assessment of market conditions to avoid false signals.

#### Hybrid Strategy:

- **Combine**: Use both conservative and contrarian strategies based on confidence thresholds.
- **Action**: Set a higher threshold for buying (only buy when the predicted probability of price increase is significantly high) and a lower threshold for selling or staying out of the market.
- **Risk Management**: Adjust thresholds dynamically based on recent model performance and market volatility.
